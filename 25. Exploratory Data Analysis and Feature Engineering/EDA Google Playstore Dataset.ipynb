```python
df_copy = df.copy()
```

```python
df_copy.drop(df_copy.index[10472],inplace=True)
```

```python
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
df['Reviews'] = df_copy['Reviews'].astype(int)
```

```python
df_copy['Size'].unique()
```

```python
df_copy['Size'] = df_copy['Size'].str.replace('M','000')
df_copy['Size'] = df_copy['Size'].str.replace('k','')
df_copy['Size'] = df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size'] = df_copy['Size'].astype(float)
```

```python
df_copy['Size']
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
chars_to_move =  ['+', '$', ',']
cols_to_clean = ['Installs', 'Price']
for col in cols_to_clean:
    for char in chars_to_move:
        df_copy[col] = df_copy[col].str.replace(char, '')
    
```

```python
df_copy['Installs'].unique()
```

```python
df_copy['Price'].unique()
```

```python
df_copy['Installs'] = df_copy['Installs'].astype(int)
df_copy['Price'] = df_copy['Price'].astype(float)
```

```python
df_copy.info()
```

```python
df_copy['Last Updated'].unique()
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.info()
```

```python
df_copy.to_csv('google_cleaned.csv')
```

```python
df_copy = pd.read_csv('google_cleaned.csv', index_col=0)
df_copy.head()
```

```python
df_copy[df_copy.duplicated('App')].shape
```

```python
df_copy.drop_duplicates(subset='App', keep='first', inplace=True)
```

```python
df_copy.shape
```

```python
numeric_features = [
    feature for feature in df_copy.columns if df_copy[feature].dtype != "O"
]
categorical_features = [
    feature for feature in df_copy.columns if df_copy[feature].dtype == "O"
]
print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
print('\nWe have {} categorical features : {}'.format(len(categorical_features), categorical_features))
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
## Proportion of count data on categorical features
for col in categorical_features:
    print(df_copy[col].value_counts(normalize=True)*100)
    print('-------------------------')
```

```python
## Proportion of count data on numerical features
plt.figure(figsize=(15, 15))
plt.suptitle('Univariate Analysis of Numerical Features',fontsize=20,fontweight='bold',alpha = 0.8 , y =1.)

for i in range(len(numeric_features)):
    plt.subplot(5, 3, i+1)
    sns.kdeplot(x=df_copy[numeric_features[i]], shade=True, color='r')
    plt.xlabel(numeric_features[i])
    plt.tight_layout()
```

### Observations:
#### Rating and Year is left skewed
#### Price and Size,Installs is right skewed

```python
## categorical columns
plt.figure(figsize=(20, 15))
plt.suptitle('Univariate Analysis of Categorical Features', fontsize=20, fontweight='bold', alpha=0.8, y=1.)

category = ['Type', 'Content Rating']
for i in range(0, len(category)):
    plt.subplot(2, 2, i+1)
    sns.countplot(x=df_copy[category[i]], palette="Set2")
    plt.xlabel(category[i])
    plt.xticks(rotation=45)
    plt.tight_layout()
```
